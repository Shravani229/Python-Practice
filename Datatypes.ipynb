```python
#Built-in Datatypes
#Numeric Datatypes

#1. Int Datatype
a = 10
print(type(a))
```

```python
#2. Float Datatype
b = 10.5
print(type(b))
```

```python
#3. Complex Datatype
c = 2+4j 
print(type(c))
```

```python
e = False 
print(type(e))
```
