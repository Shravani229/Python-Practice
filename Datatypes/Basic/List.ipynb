```python
# Add item

List_01 = ["apple","banana","mango"]

List_01.append("Grapes")
List_01
```

```python
# Insert item

List_01 = ["apple","banana","mango","grapes"]
List_01.insert(1,"kiwi")
List_01
```

```python
# Combine both lists

L1 = ["apple","banana","mango","grapes"]
L2 = ["pineapple","kiwi"]
L3 = L1 + L2
L3
```

```python
# Remove specified item

L1 = ["apple","banana","mango","grapes"]
L1.remove("banana")
L1
```

```python
# Remove by index

L1 = ["apple","banana","mango","grapes"]
L1.pop(1)
L1
```

```python
# Delete by index

L1 = ["apple","banana","mango","grapes"]
del L1[0]
L1
```

```python
# Remove last item

L1 = ["apple","banana","mango","grapes"]
L1.pop()
L1
```

```python
# Clear all items

L1 = ["apple","banana","mango","grapes"]
L1.clear()
L1
```

```python
# Delete entire list

L1 = ["apple","banana","mango","grapes"]
del L1
L1
```

```python
# Combine list in other list

L1 = ['apple', 'kIWI', 'banana'] 
L2 = ['mango', 'Pineapple']

L1.append(L2)
L1

```

```python
# Copy
#Shallow Copy

L1 = [1,2,3,4,5]
L2 = L1.copy()
L2
```

```python
#Assignment / Reference

L1 = [1,2,3,4,5]
L2=L1
L2
```

```python
#Joining / Merging
#Merge with +

L1 = [1,2,3,4,5]
L2 = [6,7,8,9,10]
L3 = L1 + L2
L3
```

```python
#Extend List

L1 = [1,2,3,4,5]
L2 = [6,7,8,9,10]
L3 = L1.extend(L2)
L1
```

```python
#Sort
#Sorted Copy

L1 = [6,5,8,7]
L2 = sorted(L1)
L2
```

```python
#Sort In Place

L1 = [5,4,6,8,3]
L1.sort()
L1
```

```python
#Reverse
#Reverse List

L1 = [1,2,3,4]
L1.reverse()
L1
```

```python
# Reverse Using Slicing

L1 = [1,2,3,4]
L1 = L1[::-1]
L1
```

```python
#Search / Count

L1 = [2,3,4,6,5]
L1.index(2)
L1
```
