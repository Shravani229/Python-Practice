```python
# Add Items

mySet = {"apple", "banana", "mango"}

mySet.add("PineApple")
mySet
```

```python
# Add Items

mySet = {"apple", "banana", "mango"}

abc = ["Pipeapple","Grapes"]
mySet.update(abc)
mySet
```

```python
# Remove an item

new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.remove("banana")
new_set
```

```python
# Remove an item

new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.discard("banana")
new_set
```

```python
# Remove an item

new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.pop()
new_set
```

```python
# Empty and delete the set

new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.clear()
new_set
```

```python
del new_set
new_set
```

```python
# Set Operations
# Union()

set1 = {1,2,3}
set2 = {4,5,6}
set1.union(set2)
```

```python
# Update a set

set1 = {1,2,3}
set2 = {4,5,6}
set1.update(set2)
set1
```

```python
# Union
set1 = {1, 2, 3}
set2 = {3, 4, 5}

set3 = set1.union(set2)

set3
```
