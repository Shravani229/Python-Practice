```python
# Access Items

myTuple = ("apple", "banana", "mango")
myTuple[-1]
```

```python
# Slicing

myTuple = ("apple", "banana", "mango", "grapes")
myTuple[:4]
```

```python
#Search / Count
#index()

myTuple = ("apple", "banana", "mango", "grapes")
myTuple.index("banana")
```

```python
# count()

myTuple = ("apple", "banana", "apple")
myTuple.count("apple")
```

```python
# Add / Join
# +

tuple1 = ("apple", "banana")
tuple2 = ("mango", "orange")
newTuple = tuple1 + tuple2
newTuple
```

```python
# Check Item
# in

myTuple = ("apple", "banana", "mango", "grapes")
"banana" in myTuple
```

```python
# Length
# len()

myTuple = ("apple", "banana", "mango", "grapes")
len(myTuple)
```

```python
# Copy
# Assignment

myTuple = ("apple", "banana", "mango", "grapes")
newTuple = myTuple
newTuple
```

```python
# sorted

myTuple = ("apple", "banana", "mango", "grapes")
newTuple = sorted(myTuple)
newTuple
```
