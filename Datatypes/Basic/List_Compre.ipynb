```python
# [Exp for i in iterable if_condition]
# [Exp loop if_condiion]
# Print even numbers

[i for i in range(1,10) if i%2==0]
```

```python
# Print square of even numbers

[i**2 for i in range(1,10) if i%2==0]
```

```python
newlist = []
for i in range(15):
    if i<5:
        newlist.append(i*2)
    elif i<10:
        newlist.append(1/2)
    else:
        newlist.append(i+2) 
newlist               
```

```python
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range (15)]   
```

```python
# Dictionary Comprehension
# [i*2 if i<5 else i/2 if i<10 else i+2 for i in range (15)]

{i: i*2 if i<5 else i/2 if i<10 else i+2 for i in range (15)}
```
