```python
# Copy items

dict1 = {"name":"Shravani","age":21}
newdict2 = dict1.copy()
newdict2
```

```python
# Merge two dictionaries

dict1 = {"name":"Shravani","age":21}
dict2 = {"Pin":414001,"Area":"Katraj"}

newdict = {**dict1,**dict2}
newdict
```

```python
newdict1 = dict(dict1,**dict2)
newdict1
```

```python
# Join

dict1 = {"name":"Shravani","age":21}
dict2 = {"Pin":414001,"Area":"Katraj"}
dict1.update(dict2)
dict1
```

```python
# Reverse 

myDict = {"name": "John", "age": 25, "city": "Pune"}

myDict

dict(reversed(list(myDict.items())))
```

```python
# Sorting
myDict = {"name": "John", "age": 25, "city": "Pune"}

newDict = dict(sorted(myDict.items()))
newDict
```
