```python
# list_02 = [1,2,3]
# dict comprehension

def printSquares(list_02):
    newDict_12 = {}

    for i in list_02:
        newDict_12[i] = i*i
    print(newDict_12)
printSquares(list_02) 
```

```python
list_02 = [1, 2, 3]

def printSquares(list_02):
    newDict_12 = {i: i*i for i in list_02}
    print(newDict_12)

printSquares(list_02)
```

```python
# save values in new list which are less than 5

list1 = [12,2,10,11,5,3]

def printValuesLessThan5(list1):
    list2 = [x for x in list1 if x < 5]
    print(list2)
printValuesLessThan5(list1)
```
