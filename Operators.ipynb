```python
# Operators

# 1. Arithmetic Operators
# Addition Operator

x = 10
y = 20
z = x+y
print("Answer : ", z)
```

```python
# 2. Assignment Operator
# Equalto Operator
x = 5
print("X :",x)

# Addition and Equalto Operator
x = 5
x+=5
print("X : ",x)
```

```python
# 3. Comparison Operator
# Equalto - Equalto Operator
x = 5
y = 3
print(x==y)

# Not Equalto Operator
x = 5
y = 3
print(x!=y)
```

```python
# 4. Logical Operator
# And Operator
x = 5
print(x>3 and x<3)

# Or Operator
x = 5
print(x>3 or x<3)
```

```python
# 5. Identity Operator
# Is Operator
x = ["Apple", "Mango"]
y = ["Apple", "Mango"]
z = x
x is not y

# Is Not Operator
x = ["Apple", "Mango"]
y = ["Apple", "Mango"]
z = x
x is z
```

```python
# 6. Membership Operator
x = ["Apple", "Mango"]
print("Banana" in x)

print("Grapes" not in x)
```
