# Real World Example

Single Inheritance

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")

class Car(Vehicle):
    def drive(self):
        print("Car is driving")

c = Car()
c.start()
c.drive()
```

Multiple Inheritance

```python
class Camera:
    def click_photo(self):
        print("Photo clicked")

class MusicPlayer:
    def play_music(self):
        print("Playing music")

class Smartphone(Camera, MusicPlayer):
    def call(self):
        print("Making a call")

s = Smartphone()
s.click_photo()
s.play_music()
s.call()
```

Multi Level Inheritance

```python
class Person:
    def introduce(self):
        print("I am a person")

class Student(Person):
    def study(self):
        print("I am studying")

class CollegeStudent(Student):
    def attend_lecture(self):
        print("Attending college lecture")

c = CollegeStudent()
c.introduce()
c.study()
c.attend_lecture()
```
