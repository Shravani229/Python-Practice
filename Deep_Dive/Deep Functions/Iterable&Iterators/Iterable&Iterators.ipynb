```python
# Python Iterators

'''Apply Filter'''

FruitsList = [1,2,3,4,5,6,7,8]
def extractOnlyEvenNumber(i):
    return i % 2 == 0
abc = filter(extractOnlyEvenNumber,FruitsList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```

```python
# Iterable

fruitslist = ["apple","banana","cherry","orange"]
print(dir(fruitslist))

for i in fruitslist:
    print(i)
```

```python
# Create an iterator from iterable

mystr = "banana"
myit = iter(mystr)
print(myit)
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```
