```python
# Case Change
# Uppercase
# Lowercase
# Capitalize

txt = "Hello My Friends"
txt.upper()
txt.lower()
txt.capitalize()
```

```python
# Replace

txt = "I like bananas"
txt.replace("bananas","apples")
```

```python
# Join()

my_list01 = ["John", "Peter", "Vicky"]
"".join(my_list01)
```

```python
# Split

txt = "welcome to the jungle"
txt.split()
```

```python
# Count()

txt = "I love apples, apples are my favourite fruit"
txt.count("apples")
```

```python
# Index()

txt = "Hello, welcome to my world."
txt.index("welcome")

```

```python
# Startswith()

txt = "Hello, welcome to my world."
txt.startswith("Hello")
```

```python
# Endswith()

txt = "Hello, welcome to my world."
txt.endswith(".")
```

```python
# find()

txt = "Hello, welcome to my world."
txt.find("welcome")
```

```python
# isnumeric()

txt = "456573"
txt.isnumeric()
```

```python
# isdigit()

txt = "478593"
txt.isdigit()
```

```python
# isalpha()

txt = "Shravani"
txt.isalpha()
```

```python
# extract substring

new_txt = "rajesh40"
new_txt[0:6] # used slicing

new_txt = "rajesh40"
new_str = ""
for i in new_txt:
    if i.isalpha() == True:
        new_str = new_str + i
print(new_str)    
```
