```python
# Conditional Statement
# IF Condition



i = 3

if i <5:
    print("Values is less than 5")
elif i < 7:
    print("Values is less than 7")
elif i < 10:
    print("Values is less than 10")
else:
    print("-----------")
```

```python
# For Loop
# range
for i in range(0,10):
    print(i)
```

```python
# While Loop
# print number from 0 to 10

i = 0

while i<=10:
    print(i)
    i = i + 1
```
